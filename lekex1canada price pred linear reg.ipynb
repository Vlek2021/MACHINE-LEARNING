```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv ('homeprices.csv')
```

```python
df.head(4)
```

```python
%matplotlib inline
plt.xlabel ('area')
plt.ylabel ('price')
plt.scatter (df.area, df.price, color = 'green', marker = '+')
```

```python
new_df = df.drop('price', axis='columns')  ###tryign to get the independent variable, so we have to drop the other column
new_df
```

```python
price = df.drop('area',axis = 'columns')    ###tryign to get the independent variable, so we have to drop the other column
price
```

```python
reg= linear_model.LinearRegression()
reg.fit(new_df,price)
```

```python
reg.predict([[2000]])
```
