```python
import pandas as pd
import numpy as np
from sklearn import linear_model
reg =linear_model.LinearRegression()
```

```python
df=pd.read_csv('hiring.csv')
df
```

```python
df['experience'] = df['experience'].fillna("zero")
df
```

```python
df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(df['interview_score(out of 10)'].median())
```

```python
df
```

```python
!pip install word2number
from word2number import w2n
###df['experience'] = df['experience'].astype(str)####
df['experience']= df['experience'].apply(w2n.word_to_num)
```

```python
df
```

```python
ab= df.drop ('salary($)',axis='columns')
ab
```

```python
# Returns a new DataFrame without these columns
cd = df.drop(columns=['experience', 'test_score(out of 10)','interview_score(out of 10)'])
cd
```

```python
reg.fit(ab,cd)
```

```python
reg.intercept_
```

```python
reg.predict([[2,9,6]]) ###predicting the salary of an applicant with an expericene of 2 years, 9 test score and 6 interview score
```

```python
reg.predict([[12,10,10]]) ###predicting the salary of an applicant with an expericene of 12 years, 10 test score and 10 interview score
```
